```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
df = df.replace([np.inf, -np.inf], np.nan)
```

```python
numerical_columns = df.select_dtypes(include=np.number).columns

for col in numerical_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
print(missing[missing > 0])
```

```python
categorical_columns = df.select_dtypes(include="object").columns

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

## Key Insights

### Insight 1: Kharif is the best-performing season

Kharif recorded the highest average yield of **5.63 tonnes/ha** and the highest average profit of **₹1,78,914.65**. It also achieved better water efficiency than Rabi and Zaid. However, Kharif had the highest disease and pest risk at **54.47%**.

### Insight 2: Zaid has the weakest economic performance

Zaid recorded the lowest average yield of **4.64 tonnes/ha** and an average loss of **₹24,804.82**. It also consumed the highest average water of **6,419.89 m³**, while having the lowest water efficiency of **4.41 tonnes/1000 m³**.

### Insight 3: Sugarcane and Chilli are the most profitable crops

Sugarcane generated the highest average profit of **₹8,17,187.99**, followed by Chilli with **₹7,50,878.34**. Sugarcane also showed the highest water efficiency among the crops analyzed.

### Insight 4: Drip irrigation shows strong overall performance

Drip irrigation recorded the highest average yield of **6.58 tonnes/ha** and the highest average profit of **₹2,19,626** among the irrigation methods analyzed.

### Insight 5: Water efficiency has a strong relationship with yield

Water efficiency showed a strong positive correlation of **0.913 with yield**, indicating that efficient utilization of water is strongly associated with higher agricultural productivity in this dataset.

### Insight 6: Revenue has a strong relationship with profit

Revenue showed a strong positive correlation of **0.887 with profit**, indicating that revenue generation is one of the major factors associated with agricultural profitability.

### Insight 7: Some agricultural inputs show weak linear relationships

Fertilizer quantity and seed quality showed very weak linear correlations with yield in this dataset. This indicates that simply increasing these inputs does not necessarily correspond to higher yield based on the observed linear relationships.

##  Recommendations

1. **Prioritize suitable crops during high-performing seasons**, particularly Kharif where overall productivity and profitability were strongest.

2. **Improve Zaid-season planning** by carefully managing irrigation, crop selection and production costs.

3. **Consider profitable crops such as Sugarcane and Chilli** where their resource requirements and local conditions are suitable.

4. **Promote efficient irrigation practices**, particularly drip irrigation, to improve productivity and profitability.

5. **Improve water-use efficiency** because it showed a strong relationship with yield.

6. **Monitor disease and pest risks**, especially during Kharif, and adopt appropriate preventive crop-management practices.

7. **Use data-driven input planning** instead of increasing fertilizer or other inputs without considering their actual relationship with productivity.

---

##  Conclusion

The Seasonal Agriculture Performance Analysis shows that agricultural performance varies considerably across seasons, crops and irrigation methods. **Kharif emerged as the strongest season overall**, while **Zaid showed comparatively weaker economic performance**.

The analysis also identified **Sugarcane and Chilli as highly profitable crops** and showed that **Drip irrigation achieved strong yield and profit performance**. Water efficiency had a particularly strong relationship with yield, while revenue was strongly associated with profit.

Overall, the analysis demonstrates how agricultural data can be used to compare seasonal performance, understand resource utilization, identify profitable crop combinations and support **data-driven agricultural planning and decision-making**.
